```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

# Common Seaborn Chart Types

| Chart Type      | Function                      | Purpose                                    | Best For                        |
| --------------- | ----------------------------- | ------------------------------------------ | ------------------------------- |
| Scatter Plot    | `sns.scatterplot()`           | Relationship between two numeric variables | Correlation, clusters           |
| Line Plot       | `sns.lineplot()`              | Trends over continuous values/time         | Time series                     |
| Bar Plot        | `sns.barplot()`               | Compare category averages                  | Category comparisons            |
| Count Plot      | `sns.countplot()`             | Count observations in categories           | Frequency analysis              |
| Histogram       | `sns.histplot()`              | Distribution of one variable               | Data distribution               |
| KDE Plot        | `sns.kdeplot()`               | Smooth distribution curve                  | Probability density             |
| ECDF Plot       | `sns.ecdfplot()`              | Empirical cumulative distribution          | Distribution comparison         |
| Box Plot        | `sns.boxplot()`               | Distribution with quartiles                | Outlier detection               |
| Violin Plot     | `sns.violinplot()`            | Distribution + density                     | Compare distributions           |
| Boxen Plot      | `sns.boxenplot()`             | Large dataset distribution                 | Many observations               |
| Strip Plot      | `sns.stripplot()`             | Individual observations                    | Small datasets                  |
| Swarm Plot      | `sns.swarmplot()`             | Non-overlapping individual points          | Detailed category comparison    |
| Point Plot      | `sns.pointplot()`             | Mean with confidence intervals             | Compare group means             |
| Error Bar Plot  | `sns.pointplot(errorbar=...)` | Mean + error bars                          | Statistical summaries           |
| Joint Plot      | `sns.jointplot()`             | Relationship + marginal distributions      | Bivariate analysis              |
| Pair Plot       | `sns.pairplot()`              | Relationships among multiple variables     | Exploratory Data Analysis (EDA) |
| Heatmap         | `sns.heatmap()`               | Matrix visualization                       | Correlation matrix              |
| Clustermap      | `sns.clustermap()`            | Heatmap with hierarchical clustering       | Cluster analysis                |
| Rug Plot        | `sns.rugplot()`               | Tick marks for observations                | Distribution visualization      |
| Regression Plot | `sns.regplot()`               | Scatter + regression line                  | Linear relationships            |
| LM Plot         | `sns.lmplot()`                | Regression with faceting                   | Grouped regression              |
| Residual Plot   | `sns.residplot()`             | Residual analysis                          | Regression diagnostics          |
| Displot         | `sns.displot()`               | Figure-level distribution plot             | Histograms, KDE, ECDF           |
| Relplot         | `sns.relplot()`               | Figure-level relational plot               | Scatter/Line with facets        |
| Catplot         | `sns.catplot()`               | Figure-level categorical plots             | Box, violin, bar, etc.          |
| FacetGrid       | `sns.FacetGrid()`             | Multiple plots by category                 | Multi-panel visualization       |
| PairGrid        | `sns.PairGrid()`              | Custom pairwise plots                      | Advanced EDA                    |
| JointGrid       | `sns.JointGrid()`             | Custom joint distributions                 | Advanced bivariate analysis     |


# Common Seaborn Parameters

| Parameter   | Description                                     | Used In                                              |
| ----------- | ----------------------------------------------- | ---------------------------------------------------- |
| `data`      | DataFrame or dataset                            | Almost all plots                                     |
| `x`         | X-axis variable                                 | Almost all plots                                     |
| `y`         | Y-axis variable                                 | Most plots                                           |
| `hue`       | Group by color                                  | Most plots                                           |
| `style`     | Marker/line style                               | Scatter, line                                        |
| `size`      | Marker size                                     | Scatter                                              |
| `palette`   | Color palette                                   | Most plots                                           |
| `color`     | Single plot color                               | Most plots                                           |
| `markers`   | Marker style(s)                                 | Scatter, line                                        |
| `marker`    | Single marker                                   | Scatter                                              |
| `linewidth` | Line width                                      | Line, regression                                     |
| `linestyle` | Line style                                      | Line plots                                           |
| `alpha`     | Transparency                                    | Most plots                                           |
| `fill`      | Fill area under curve                           | KDE, histogram                                       |
| `bins`      | Number of histogram bins                        | Histogram                                            |
| `binwidth`  | Width of bins                                   | Histogram                                            |
| `multiple`  | Layer/stack/dodge histograms                    | Histogram                                            |
| `stat`      | Statistic to display (`count`, `density`, etc.) | Histogram                                            |
| `kde`       | Overlay KDE curve                               | Histogram                                            |
| `orient`    | Horizontal/vertical orientation                 | Box, violin, bar                                     |
| `order`     | Category order                                  | Categorical plots                                    |
| `hue_order` | Hue category order                              | Grouped plots                                        |
| `estimator` | Aggregation function (mean, median, etc.)       | Bar, point - available in barplot, lineplot, pointplot only                                         |
| `errorbar`  | Error bar type                                  | Bar, point                                           |
| `ci`        | Confidence interval (older versions)            | Regression, point                                    |
| `capsize`   | Error bar cap size                              | Bar, point                                           |
| `dodge`     | Separate grouped elements                       | Bar, violin, box                                     |
| `jitter`    | Spread overlapping points                       | Strip plot                                           |
| `split`     | Split violins by hue                            | Violin plot                                          |
| `inner`     | Interior representation in violin               | Violin plot                                          |
| `whis`      | Whisker length                                  | Box plot                                             |
| `showmeans` | Show mean value                                 | Box plot                                             |
| `annot`     | Display values inside cells                     | Heatmap                                              |
| `fmt`       | Annotation format                               | Heatmap                                              |
| `cmap`      | Colormap                                        | Heatmap                                              |
| `center`    | Center value for colormap                       | Heatmap                                              |
| `vmin`      | Minimum color value                             | Heatmap                                              |
| `vmax`      | Maximum color value                             | Heatmap                                              |
| `mask`      | Hide selected cells                             | Heatmap                                              |
| `robust`    | Robust color scaling                            | Heatmap                                              |
| `corr`      | Correlation method (user-computed matrix)       | Heatmap workflow                                     |
| `kind`      | Plot type (`scatter`, `line`, `hist`, etc.)     | `relplot()`, `catplot()`, `jointplot()`, `displot()` |
| `col`       | Facet by columns                                | Figure-level plots                                   |
| `row`       | Facet by rows                                   | Figure-level plots                                   |
| `col_wrap`  | Wrap facet columns                              | Figure-level plots                                   |
| `height`    | Height of each facet                            | Figure-level plots                                   |
| `aspect`    | Width-to-height ratio                           | Figure-level plots                                   |
| `legend`    | Show legend                                     | Most plots                                           |
| `ax`        | Existing Matplotlib axes                        | Axes-level plots                                     |


```python
df=pd.read_excel("/Users/kanchana/Desktop/DATASET/financial_loan.xlsx")
```

```python
df.columns
```

```python
state=['DC','AK','HI']
filtered_df=df[df['address_state'].isin(state)]
```

```python
grouped_df=filtered_df.groupby(['address_state','grade'])['loan_amount'].mean() #order matters here
grouped_df
```

```python
grouped_df.unstack()
```

```python
grouped_df.unstack().plot(kind='bar')
plt.show()
```

```python
sns.barplot(filtered_df,x='address_state', y='loan_amount')
```

```python
# using seaborn 
sns.barplot(filtered_df,x='address_state', y='loan_amount',hue='grade')
```

```python
sns.barplot(filtered_df,x='address_state', y='loan_amount',hue='grade', estimator=np.mean)

# estimator to change the aggregation of loan amount

# Since Seaborn is built on top of matplotlib, we can use matplotlib and seaborn together

plt.xlabel('state')
plt.ylabel('mean loan amount')
plt.title('Loan amount by state and grade')
plt.legend(title='Grade',bbox_to_anchor=(1,1))
plt.show()
```

# Common bbox_to_anchor Positions

| `bbox_to_anchor` | Position                | Typical `loc`    |
| ---------------- | ----------------------- | ---------------- |
| `(1, 1)`         | Outside upper-right     | `"upper left"`   |
| `(1, 0.5)`       | Outside right-center    | `"center left"`  |
| `(1, 0)`         | Outside lower-right     | `"lower left"`   |
| `(0, 1)`         | Upper-left corner       | `"upper left"`   |
| `(0.5, 1)`       | Top-center              | `"lower center"` |
| `(0.5, 0)`       | Bottom-center           | `"upper center"` |
| `(0, 0.5)`       | Left-center             | `"center right"` |
| `(0, 0)`         | Bottom-left             | `"lower left"`   |
| `(1.2, 1)`       | Far outside upper-right | `"upper left"`   |
| `(1.05, 0.5)`    | Slightly outside right  | `"center left"`  |
| `(-0.2, 1)`      | Outside upper-left      | `"upper right"`  |
| `(0.5, 1.15)`    | Above the plot          | `"lower center"` |
| `(0.5, -0.15)`   | Below the plot          | `"upper center"` |


# Common loc Values

| `loc`            | Legend Anchor Point  |
| ---------------- | -------------------- |
| `"best"`         | Automatically chosen |
| `"upper right"`  | Upper-right corner   |
| `"upper left"`   | Upper-left corner    |
| `"lower left"`   | Lower-left corner    |
| `"lower right"`  | Lower-right corner   |
| `"right"`        | Center-right         |
| `"center left"`  | Center-left          |
| `"center right"` | Center-right         |
| `"lower center"` | Bottom-center        |
| `"upper center"` | Top-center           |
| `"center"`       | Exact center         |


```python
sns.set_theme(style='darkgrid')

sns.barplot(filtered_df,x='address_state', y='loan_amount',hue='grade', estimator=np.mean)

# estimator to change the aggregation of loan amount

# Since Seaborn is built on top of matplotlib, we can use matplotlib and seaborn together

plt.xlabel('state')
plt.ylabel('mean loan amount')
plt.title('Loan amount by state and grade') 
plt.legend(title='Grade',loc="center right", bbox_to_anchor=(1.2, 0.5))# loc and bbox
plt.show()
```

```python
sns.set_theme(style='darkgrid') #background grid

sns.barplot(filtered_df,x='address_state', y='loan_amount',hue='grade', estimator=np.mean, palette='Blues') # adding palette here

# estimator to change the aggregation of loan amount

# Since Seaborn is built on top of matplotlib, we can use matplotlib and seaborn together

plt.xlabel('state')
plt.ylabel('mean loan amount')
plt.title('Loan amount by state and grade') 
plt.legend(title='Grade',loc="center right", bbox_to_anchor=(1.2, 0.5))# loc and bbox
plt.show()
```

```python
sns.set_theme(style='white', palette=None)
sns.boxplot(df, y='int_rate') # just creating a vertical boxplot

# adding customization here
plt.title('Distribution of Interest Rate',fontsize=16,fontweight='bold')
plt.ylabel('Interest Rate', fontsize=12)
sns.despine(bottom=True)
plt.show()
```

```python
sns.set_theme(style='white', palette=None)
sns.boxplot(df, x='grade',y='int_rate',order=sorted(df['grade'].unique())) # sorting and including x and y

# adding customization here
plt.title('Distribution of Interest Rate',fontsize=16,fontweight='bold')
plt.ylabel('Interest Rate', fontsize=12)
sns.despine(bottom=True) #USED TO REMOVE BOUNDARIES
plt.show()
```

```python
sns.set_theme(style='white', palette=None)
sns.boxplot(df, x='grade',y='int_rate',order=sorted(df['grade'].unique()),hue='grade', palette='coolwarm') # sorting and including x and y

# adding customization here
plt.title('Distribution of Interest Rate',fontsize=16,fontweight='bold')
plt.ylabel('Interest Rate', fontsize=12)
sns.despine(bottom=True) #USED TO REMOVE BOUNDARIES
plt.show()
```

```python
plt.figure(figsize=(16,10))
sns.histplot(df,x='loan_amount',binwidth=1000,kde=True)
plt.xlabel('loan_amount')
plt.ylabel('')
plt.title('Loan amount distribution', fontsize=19 ) 
plt.grid(axis='y', linestyle='--')
plt.show()
```
