# Common Seaborn Chart Types

| Chart Type      | Function                      | Purpose                                    | Best For                        |
| --------------- | ----------------------------- | ------------------------------------------ | ------------------------------- |
| Scatter Plot    | `sns.scatterplot()`           | Relationship between two numeric variables | Correlation, clusters           |
| Line Plot       | `sns.lineplot()`              | Trends over continuous values/time         | Time series                     |
| Bar Plot        | `sns.barplot()`               | Compare category averages                  | Category comparisons            |
| Count Plot      | `sns.countplot()`             | Count observations in categories           | Frequency analysis              |
| Histogram       | `sns.histplot()`              | Distribution of one variable               | Data distribution               |
| KDE Plot        | `sns.kdeplot()`               | Smooth distribution curve                  | Probability density             |
| ECDF Plot       | `sns.ecdfplot()`              | Empirical cumulative distribution          | Distribution comparison         |
| Box Plot        | `sns.boxplot()`               | Distribution with quartiles                | Outlier detection               |
| Violin Plot     | `sns.violinplot()`            | Distribution + density                     | Compare distributions           |
| Boxen Plot      | `sns.boxenplot()`             | Large dataset distribution                 | Many observations               |
| Strip Plot      | `sns.stripplot()`             | Individual observations                    | Small datasets                  |
| Swarm Plot      | `sns.swarmplot()`             | Non-overlapping individual points          | Detailed category comparison    |
| Point Plot      | `sns.pointplot()`             | Mean with confidence intervals             | Compare group means             |
| Error Bar Plot  | `sns.pointplot(errorbar=...)` | Mean + error bars                          | Statistical summaries           |
| Joint Plot      | `sns.jointplot()`             | Relationship + marginal distributions      | Bivariate analysis              |
| Pair Plot       | `sns.pairplot()`              | Relationships among multiple variables     | Exploratory Data Analysis (EDA) |
| Heatmap         | `sns.heatmap()`               | Matrix visualization                       | Correlation matrix              |
| Clustermap      | `sns.clustermap()`            | Heatmap with hierarchical clustering       | Cluster analysis                |
| Rug Plot        | `sns.rugplot()`               | Tick marks for observations                | Distribution visualization      |
| Regression Plot | `sns.regplot()`               | Scatter + regression line                  | Linear relationships            |
| LM Plot         | `sns.lmplot()`                | Regression with faceting                   | Grouped regression              |
| Residual Plot   | `sns.residplot()`             | Residual analysis                          | Regression diagnostics          |
| Displot         | `sns.displot()`               | Figure-level distribution plot             | Histograms, KDE, ECDF           |
| Relplot         | `sns.relplot()`               | Figure-level relational plot               | Scatter/Line with facets        |
| Catplot         | `sns.catplot()`               | Figure-level categorical plots             | Box, violin, bar, etc.          |
| FacetGrid       | `sns.FacetGrid()`             | Multiple plots by category                 | Multi-panel visualization       |
| PairGrid        | `sns.PairGrid()`              | Custom pairwise plots                      | Advanced EDA                    |
| JointGrid       | `sns.JointGrid()`             | Custom joint distributions                 | Advanced bivariate analysis     |

# Stack complementary charts on top of each other

```python
plt.figure(figsize=(16,10))
sns.histplot(df,x='loan_amount',bins=15,kde=True) #aggregated plot
sns.rugplot(df,x='loan_amount',color='red') #disaggregated plot with individual data points
plt.xlabel('loan_amount')
plt.ylabel('')
plt.title('Loan amount distribution', fontsize=19 ) 
plt.grid(axis='y', linestyle='--')
plt.show()
```

```python
plt.figure(figsize=(12,9))
sns.set_theme(style='white', palette=None)
sns.boxplot(df, x='grade',y='int_rate',order=sorted(df['grade'].unique()),hue='grade') # sorting and including x and y
sns.stripplot(df,x='grade', y='int_rate',jitter=0.35,size=1,alpha=0.2,color='black')
# adding customization here
plt.title('Distribution of Interest Rate',fontsize=16,fontweight='bold')
plt.ylabel('Interest Rate', fontsize=12)
sns.despine(bottom=True) #USED TO REMOVE BOUNDARIES
plt.show()
```

The chart demonstrates that loan grade is a strong predictor of interest rate. Borrowers with better credit quality (Grade A) receive the lowest interest rates, whereas borrowers with poorer credit quality (Grade G) pay substantially higher rates. Additionally, the variability in interest rates tends to increase as loan grade declines, indicating more diverse pricing among higher-risk loans.

This is exactly the pattern expected in lending: higher credit risk leads to higher interest rates.

Grades A and B have relatively narrow boxes and shorter whiskers, indicating less variation in interest rates.
Grades E, F, and G exhibit wider boxes and longer whiskers, suggesting greater variability in the rates offered.

A few unusually low interest rates appear in grades B–E (around 6%), shown as points below the whiskers. These represent observations that are much lower than the typical rates for those grades.
