# **Random assignment functions in Python, Pandas, PySpark and SQL**

Random assignment means **randomly allocating records, people, customers, employees or transactions into different groups**.

For example:

- Randomly divide 1,000 customers into 5 groups.
- Assign employees randomly to teams.
- Split a dataset into 80% training and 20% testing.
- Randomly assign customers to control and treatment groups.
- Select a random sample of records.
- Assign random numbers or IDs to each row.

There are different functions depending on what you need. The most important distinction is between:

1. **Random number generation** – generate random values.
2. **Random row selection** – select random records.
3. **Random group assignment** – assign records to groups.
4. **Random sampling** – select a percentage or number of records.
5. **Randomized train-test splitting** – divide datasets.
6. **Reproducible randomization** – get the same random result again using a seed.

## **B. Randomly select rows**

```
df.sample(n=3)
```

Selects 3 random rows.

Select 50% of the dataset:

```
df.sample(frac=0.5)
```

Select with replacement:

```
df.sample(n=10, replace=True)
```

Reproducible random selection:

```
df.sample(n=3, random_state=42)
```

**When not to use&#x20;**`sample()`**:** If you need to assign every row to a group. It only selects a sample. Use random group assignment instead.

## **C. Random assignment into teams**

Assign each employee randomly to one of three teams:

```
np.random.seed(42)df['Team'] = np.random.choice(    ['Team A', 'Team B', 'Team C'],    size=len(df))
```

Example output (assignments depend on the seed and NumPy version):

| Employee | Team   |
| -------- | ------ |
| A        | Team C |
| B        | Team A |
| C        | Team C |
| D        | Team C |
| E        | Team A |
| F        | Team A |
| G        | Team C |
| H        | Team B |

This gives random assignment but does not guarantee equal team sizes.

## **D. Random assignment with approximately equal group sizes**

```
df = df.sample(frac=1, random_state=42).copy()df['Group'] = (    np.arange(len(df)) % 3 + 1)
```

This creates three groups with sizes differing by at most one.

For 8 employees, the group sizes will be 3, 3 and 2.

**Why use this instead of&#x20;**`np.random.choice()`**?**

- `np.random.choice()` can produce uneven group sizes.
- Shuffle + modulo produces nearly equal group sizes.

## **E. Assign exactly equal-sized groups**

If the number of rows is divisible by the number of groups:

```
df = df.sample(frac=1, random_state=42).copy()df['Group'] = np.repeat(    [1, 2, 3, 4],    len(df) // 4)
```

For datasets not divisible by the number of groups, use:

```
df = df.sample(frac=1, random_state=42).copy()df['Group'] = (    np.arange(len(df)) % 4 + 1)
```

This guarantees group sizes differ by at most one.

## **F. Random assignment with weighted probabilities**

```
df['Group'] = np.random.choice(    ['Control', 'Treatment'],    size=len(df),    p=[0.7, 0.3])
```

This assigns approximately:

- 70% Control
- 30% Treatment

It does not guarantee exactly 70/30.

# **3. PySpark – random assignment functions**

PySpark is especially useful for random assignment in large datasets because it supports distributed randomization.

Import:

```
from pyspark.sql import functions as Ffrom pyspark.sql.window import Window
```

Sample DataFrame:

```
data = [    (1, 'A'),    (2, 'B'),    (3, 'C'),    (4, 'D'),    (5, 'E'),    (6, 'F'),    (7, 'G'),    (8, 'H')]df = spark.createDataFrame(    data,    ['ID', 'Employee'])
```

## **A. Important PySpark random functions**

| No. | Function                        | Purpose                                     | Example                            |
| --- | ------------------------------- | ------------------------------------------- | ---------------------------------- |
| 1   | `rand()`                        | Generate random decimal values              | `F.rand()`                         |
| 2   | `rand(seed)`                    | Generate seeded random values               | `F.rand(42)`                       |
| 3   | `randn()`                       | Generate normally distributed random values | `F.randn()`                        |
| 4   | `sample()`                      | Randomly sample rows                        | `df.sample(0.5)`                   |
| 5   | `sampleBy()`                    | Stratified random sampling                  | `df.sampleBy(...)`                 |
| 6   | `orderBy(rand())`               | Randomly shuffle rows                       | `df.orderBy(F.rand())`             |
| 7   | `row_number()`                  | Assign sequential row numbers               | `F.row_number().over(w)`           |
| 8   | `ntile()`                       | Divide ordered rows into buckets            | `F.ntile(3).over(w)`               |
| 9   | `partitionBy()`                 | Define window partitions                    | `Window.partitionBy('Department')` |
| 10  | `monotonically_increasing_id()` | Generate distributed unique IDs             | `F.monotonically_increasing_id()`  |

### **1.&#x20;**`rand()`

```
df.withColumn(    'random_number',    F.rand()).show()
```

Generates random values between 0 and 1.

With seed:

```
df.withColumn(    'random_number',    F.rand(42))
```

Use: Generate random numbers or establish a random ordering.

Don't use: When you need exact equal-sized groups by itself. Combine with `row_number()` or `ntile()`.

### **2.&#x20;**`randn()`

```
df.withColumn(    'random_number',    F.randn(42)).show()
```

Generates values from a standard normal distribution (mean approximately 0 and standard deviation approximately 1).

Use: Simulations and statistical applications.

Don't use: When you need uniform random group selection. Use `rand()`.

### **3.&#x20;**`sample()`

```
df.sample(    withReplacement=False,    fraction=0.5,    seed=42)
```

Selects approximately 50% of rows.

Important: It does not guarantee an exact number of records.

With replacement:

```
df.sample(    withReplacement=True,    fraction=0.5,    seed=42)
```

Use: Random dataset sampling.

Don't use: When every record must receive a group assignment.

### **4.&#x20;**`sampleBy()`

Important for data science

This performs stratified sampling.

Suppose you want to randomly sample 50% of each department.

```
fractions = {    'IT': 0.5,    'HR': 0.5,    'Sales': 0.5}df.sampleBy(    'Department',    fractions,    seed=42)
```

Use: When you need sampling within categories while maintaining their representation.

Don't use: When you want to assign all records to entirely new groups.

### **5.&#x20;**`orderBy(F.rand())`

```
df.orderBy(F.rand(42)).show()
```

Randomly shuffles the rows.

Use: Random ordering before selecting or assigning groups.

Don't use: On very large datasets without considering the shuffle and sorting cost. This can be expensive.

## **B. Randomly assign records into groups**

### **Method 1: Using&#x20;**`rand()`**&#x20;and&#x20;**`when()`

```
df = df.withColumn(    'Group',    F.when(F.rand(42) < 0.33, 'A')     .when(F.rand(43) < 0.5, 'B')     .otherwise('C'))
```

This example uses separate random expressions and does not create exactly balanced groups. A better way for probability-based assignment is to generate one random column:

```
df = df.withColumn('r', F.rand(42))df = df.withColumn(    'Group',    F.when(F.col('r') < 0.33, 'A')     .when(F.col('r') < 0.66, 'B')     .otherwise('C')).drop('r')
```

This creates approximately equal groups, not exact group sizes.

### **Method 2: Using&#x20;**`ntile()`

Best for nearly equal-sized groups

```
w = Window.orderBy(F.rand(42))df = df.withColumn(    'Group',    F.ntile(3).over(w))
```

This divides the dataset into 3 nearly equal-sized groups.

For 8 records:

| ID | Employee | Group |
| -- | -------- | ----- |
| 1  | A        | 1     |
| 2  | B        | 1     |
| 3  | C        | 1     |
| 4  | D        | 2     |
| 5  | E        | 2     |
| 6  | F        | 2     |
| 7  | G        | 3     |
| 8  | H        | 3     |

The actual employee-to-group assignment is randomized.

**When to use:** You need balanced random assignment.

**When not to use:** You want independently randomized assignments with specified probabilities. Use `rand()`.

### **Method 3: Using&#x20;**`row_number()`

```
w = Window.orderBy(F.rand(42))df = df.withColumn(    'random_rank',    F.row_number().over(w))df = df.withColumn(    'Group',    ((F.col('random_rank') - 1) % 3) + 1)
```

This assigns records round-robin into 3 groups after random ordering.

Use: Balanced group allocation with an explicit randomized row number.

Don't use: If you only need to select a random subset. Use `sample()`.

### **Method 4: Random assignment within each department**

Suppose you want to divide employees into two groups separately within each department.

```
w = (    Window    .partitionBy('Department')    .orderBy(F.rand(42)))df = df.withColumn(    'Group',    F.ntile(2).over(w))
```

This is particularly useful for:

- A/B testing within customer segments.
- Randomized employee allocation within departments.
- Stratified experimental assignment.

**Important:** `ntile()` balances each department independently, not the entire dataset.

# **4. SQL – random assignment functions**

SQL syntax varies between databases, especially SQL Server, PostgreSQL, MySQL, Oracle and Spark SQL.

Here are the important functions.

## **A. Random number functions by SQL database**

| Database   | Random function       | Example                               |
| ---------- | --------------------- | ------------------------------------- |
| PostgreSQL | `RANDOM()`            | `SELECT RANDOM();`                    |
| MySQL      | `RAND()`              | `SELECT RAND();`                      |
| SQL Server | `NEWID()`             | `SELECT NEWID();`                     |
| SQL Server | `RAND()`              | `SELECT RAND();`                      |
| Oracle     | `DBMS_RANDOM.VALUE()` | `SELECT DBMS_RANDOM.VALUE FROM DUAL;` |
| Spark SQL  | `RAND()`              | `SELECT RAND();`                      |
| SQLite     | `RANDOM()`            | `SELECT RANDOM();`                    |

`NEWID()` generates a GUID rather than a numeric random decimal. It is commonly used to randomize SQL Server row ordering.

## **B. Randomly select records**

**PostgreSQL:**

```
SELECT *
FROM employees
ORDER BY RANDOM()
LIMIT 5;
```

**MySQL:**

```
SELECT *
FROM employees
ORDER BY RAND()
LIMIT 5;
```

**SQL Server:**

```
SELECT TOP 5 *
FROM employees
ORDER BY NEWID();
```

**Spark SQL:**

```
SELECT *
FROM employees
ORDER BY RAND(42)
LIMIT 5;
```

Use: Select random records.

Don't use: When every row must be assigned to a group.

## **C. Random group assignment using CASE**

PostgreSQL, MySQL and Spark SQL:

```
SELECT *,
       CASE
           WHEN RANDOM() < 0.33 THEN 'A'
           WHEN RANDOM() < 0.66 THEN 'B'
           ELSE 'C'
       END AS random_group
FROM employees;
```

For correct threshold-based probabilities, generate the random value once:

```
WITH random_data AS (
    SELECT *,
           RANDOM() AS r
    FROM employees
)
SELECT *,
       CASE
           WHEN r < 0.33 THEN 'A'
           WHEN r < 0.66 THEN 'B'
           ELSE 'C'
       END AS random_group
FROM random_data;
```

For MySQL and Spark SQL, replace `RANDOM()` with `RAND()`.

For SQL Server, replace it with an appropriate `NEWID()`-based ordering or another randomization strategy.

## **D. Random assignment using&#x20;**`NTILE()`

Very important SQL function

```
SELECT *,
       NTILE(3) OVER (
           ORDER BY RANDOM()
       ) AS random_group
FROM employees;
```

This creates 3 nearly equal-sized groups.

Database-specific changes:

| Database   | Random expression   |
| ---------- | ------------------- |
| PostgreSQL | `RANDOM()`          |
| MySQL      | `RAND()`            |
| SQL Server | `NEWID()`           |
| Oracle     | `DBMS_RANDOM.VALUE` |
| Spark SQL  | `RAND(42)`          |

Example for SQL Server:

```
SELECT *,
       NTILE(3) OVER (
           ORDER BY NEWID()
       ) AS random_group
FROM employees;
```

Use: Randomly split records into nearly equal groups.

Don't use: When group sizes need to follow weighted probabilities such as 70%, 20%, 10%.

## **E. Random assignment using&#x20;**`ROW_NUMBER()`

```
WITH random_data AS (
    SELECT *,
           ROW_NUMBER() OVER (
               ORDER BY RANDOM()
           ) AS rn
    FROM employees
)
SELECT *,
       ((rn - 1) % 3) + 1 AS random_group
FROM random_data;
```

This randomizes the rows and distributes them round-robin into 3 groups.

Use: Nearly equal-sized random groups with explicit sequential numbering.

## **F. Stratified random assignment using&#x20;**`PARTITION BY`

```
SELECT *,
       NTILE(2) OVER (
           PARTITION BY Department
           ORDER BY RANDOM()
       ) AS random_group
FROM employees;
```

This divides employees into two nearly equal groups within each department.

Use: When you want to preserve the representation of categories in experimental groups.

## **G. Other SQL sampling syntax**

| Function / syntax   | Database                          | Purpose                              |
| ------------------- | --------------------------------- | ------------------------------------ |
| `TABLESAMPLE`       | PostgreSQL, SQL Server and others | Random table sampling                |
| `ORDER BY RANDOM()` | PostgreSQL, SQLite                | Random ordering                      |
| `ORDER BY RAND()`   | MySQL                             | Random ordering                      |
| `ORDER BY NEWID()`  | SQL Server                        | Random ordering                      |
| `NTILE(n)`          | Most major SQL databases          | Divide rows into buckets             |
| `ROW_NUMBER()`      | Most major SQL databases          | Assign sequential numbers            |
| `PARTITION BY`      | Window functions                  | Perform assignment within categories |
| `LIMIT`             | PostgreSQL, MySQL, Spark SQL      | Restrict output rows                 |
| `TOP`               | SQL Server                        | Restrict output rows                 |

`TABLESAMPLE` is not equivalent to exact row selection or balanced assignment. Its sampling method and supported syntax differ by database.

# **7. Three important concepts you must understand**

**1. Random assignment does not mean equal assignment.**

For example:

```
np.random.choice(['A', 'B', 'C'], size=100)
```

May produce:

- A = 35
- B = 40
- C = 25

But using a shuffled sequence and modulo produces group sizes differing by at most one.

**2. Random seed means reproducibility, not permanent randomness.**

For example:

```
np.random.seed(42)
```

Allows you to reproduce the same random sequence under the same conditions.

In distributed PySpark, a seed alone does not guarantee identical row assignments if partitioning or execution changes.

**3. Random assignment and random sampling are different.**

- Sampling: Select only some records.
- Assignment: Give every record a group.
- Stratified sampling: Select records separately within categories.
- Stratified assignment: Assign records to groups separately within categories.
