# Filtering a Pandas DataFrame


Here are detailed notes on the 6 methods for filtering a pandas DataFrame as presented in the video **"How to filter a pandas DataFrame | 6 HELPFUL METHODS"**:

---

### **Dataset Overview**
The examples in the video use a dataset from 538 (`recent-grads.csv`) containing 173 rows and 21 columns. The data covers college majors, post-graduation employment numbers, and demographics.

---

### **1. Boolean Mask**
A boolean mask filters rows based on whether specified conditions evaluate to `True`.

* **Single Condition**: Pass a condition inside square brackets:
  `df[df['Major_category'] == 'Social Science']`.
* **Multiple Conditions**:
  * Wrap each condition in parentheses.
  * Use **`&`** for AND and **`|`** for OR (standard Python `and` / `or` will not work with Series).
  * Store criteria in a variable (e.g., `filter_criteria`) to keep the code clean and readable:
    ```python
    filter_criteria = (df['Major_category'] == 'Social Science') & (df['ShareWomen'] >= 0.5)
    df1 = df[filter_criteria]  # Returns 5 matching rows
    ```
* **Inverting Criteria (Opposite)**: Use the **tilde (`~`)** operator to invert the mask and return all rows where the condition is `False`:
  ```python
  df_opposite = df[~filter_criteria]  # Returns 168 rows
  ```

---

### **2. The `loc` Method (`df.loc`)**
The **`loc`** method functions similarly to boolean masking for row filtering, but provides the added advantage of returning **specific columns** alongside the filtered rows.

* **Select All Columns**: `df.loc[filter_criteria]`.
* **Select Specific Columns**: Pass column name(s) as a string or list in the second argument:
  ```python
  # Single column output
  df.loc[filter_criteria, 'Part_time']

  # Multiple columns output
  cols = ['Part_time', 'Sample_size', 'Total']
  df.loc[filter_criteria, cols]
  ```

---

### **3. The `query` Method (`df.query()`)**
The **`query`** method allows filtering using string expressions without repeating the DataFrame name.

* **Syntax**:
  ```python
  df.query("Major_category == 'Social Science' and ShareWomen >= 0.5")
  ```
* **In-Place Modification**: Supports `inplace=True` if you want to modify the original DataFrame directly rather than creating a copy.
* **Performance Note**: Timing tests show **Boolean Mask** and **`loc`** have nearly identical execution speeds, whereas **`query`** is noticeably slower.

---

### **4. The `iloc` Method (`df.iloc`)**
**`iloc`** filters data based on **integer index positions** of both rows and columns.

* **Slicing Ranges**:
  ```python
  df.iloc[0:10, 2:6]  # Rows 0-9 and columns at index positions 2, 3, 4, 5
  ```
* **Specific Indices**: Pass a list of index numbers:
  ```python
  df.iloc[0:10,]  # Rows 0-9 and specific columns at indices 2 and 6
  ```

---

### **5. `loc` with Named Indices**
When working with explicit row index labels (especially after setting a named index column via `df.set_index()`), **`loc`** allows selecting specific rows and columns by their labels.

* **Example with Named Index**:
  ```python
  df.set_index('Major', inplace=True)
  rows_to_keep = ['Petroleum Engineering', 'Mechanical Engineering']
  cols_to_keep = ['Major_category', 'Total']
  
  df.loc[rows_to_keep, cols_to_keep]  # Returns a targeted 2x2 DataFrame
  ```

---

### **6. Filtering Missing Data (`NaN` / Null Values)**
Methods for identifying, dropping, or isolating missing values in your dataset:

* **Check Missing Counts**: `df.isnull().sum()` counts missing values per column.
* **Drop All Rows with Missing Data**: `df.dropna(inplace=True)` removes any row containing missing values.
* **Filter Non-Null Rows in a Column**: Use `.notnull()` inside a boolean mask:
  ```python
  df[df['Total'].notnull()]
  ```
* **Isolate Missing Value Rows**:
  * For a single column: `df[df['Total'].isnull()]`.
  * Across the entire DataFrame (any column missing): `df[df.isnull().any(axis=1)]`.
