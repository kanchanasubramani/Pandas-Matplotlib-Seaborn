# Sorting Functions in pandas

| Function               | Sorts By                | When to Use                                                                                                                      | Example                                |
| ---------------------- | ----------------------- | -------------------------------------------------------------------------------------------------------------------------------- | -------------------------------------- |
| `sort_values()`        | Column values           | **Most common.** Sort rows based on one or more columns.                                                                         | `df.sort_values("Sales")`              |
| `sort_index()`         | Index labels            | When the row/column index is out of order and you want to reorder it.                                                            | `df.sort_index()`                      |
| `pd.Categorical()`     | Custom category order   | When values should follow a business order instead of alphabetical order (e.g., Low → Medium → High). Used with `sort_values()`. | `df["Priority"] = pd.Categorical(...)` |
| `argsort()`            | Returns sorting indices | When you need the positions that would sort the data instead of sorting directly. Useful with NumPy or `iloc`.                   | `df.iloc[cat.argsort()]`               |
| `nlargest()`           | Largest values          | When you only need the top *N* rows. Faster than sorting the entire DataFrame.                                                   | `df.nlargest(10, "Sales")`             |
| `nsmallest()`          | Smallest values         | When you only need the bottom *N* rows.                                                                                          | `df.nsmallest(5, "Profit")`            |
| `rank()`               | Assigns ranks           | When you need rankings rather than reordered rows.                                                                               | `df["Rank"] = df["Sales"].rank()`      |
| `sort_values(key=...)` | Custom sorting logic    | When you want to sort without modifying the original column.                                                                     | `df.sort_values("Segment", key=...)`   |

Internally, Pandas stores something like:

Categories:


0 -> Blue

1 -> Green

2 -> Red

Data:
2, 0, 2, 1, 0

This is much more memory-efficient than storing the full strings repeatedly.

# a fixed set of valid values

You can specify the allowed categories:

````python
grades = pd.Categorical(
    ["A", "B", "A"],
    categories=["A", "B", "C", "D"]
)

print(grades.categories)
````

# Even though "C" and "D" don't appear in the data, they are still considered valid categories.

# when to use pd.categorical vs sort_values(keys=..)

 The answer depends on **whether the custom order is temporary or part of your data model**.

## Comparison

| Aspect                                        | `pd.Categorical()` | `sort_values(key=...)`                |
| --------------------------------------------- | ------------------ | ------------------------------------- |
| Changes the column dtype                      | ✅ Yes (`category`) | ❌ No                                  |
| Custom sort order                             | ✅ Yes              | ✅ Yes                                 |
| Order persists                                | ✅ Yes              | ❌ No (only for that sort)             |
| Memory efficient                              | ✅ Yes              | ❌ No                                  |
| Faster for repeated operations                | ✅ Yes              | ❌ Recreates the categorical each time |
| Good for one-time sorting                     | 😐 Overkill        | ✅ Best choice                         |
| Good for repeated sorting, grouping, plotting | ✅ Best choice      | ❌ Not ideal                           |

---

## Option 1: `pd.Categorical()` (Persistent)

```python
order = ["Consumer", "Corporate", "Home Office"]

df["Segment"] = pd.Categorical(
    df["Segment"],
    categories=order,
    ordered=True
)

df.sort_values("Segment")
```

### What happens?

The `Segment` column **becomes a categorical column**.

Now every operation understands the order:

```python
df.sort_values("Segment")
df.groupby("Segment")
df["Segment"].min()
df["Segment"].max()
```

No need to specify the order again.

### Use when

* The column has a natural business order.
* You'll sort by it multiple times.
* You'll group, plot, or analyse using that order.

Example:

```
Priority:
Critical
High
Medium
Low
```

This order should exist everywhere, not just during one sort.

---

## Option 2: `sort_values(key=...)` (Temporary)

```python
order = ["Consumer", "Corporate", "Home Office"]

df.sort_values(
    "Segment",
    key=lambda s: pd.Categorical(s, categories=order, ordered=True)
)
```

### What happens?

* The column stays as `object` (or string).
* Pandas creates a temporary categorical **only for sorting**.
* After sorting, it discards it.

Nothing changes in your DataFrame.

### Use when

* You only need one custom sort.
* You don't want to modify the DataFrame.
* The custom order is specific to one report.

---

## Why not always use `key=`?

Suppose you sort five times:

```python
df.sort_values("Segment", key=...)
df.sort_values("Segment", key=...)
df.sort_values("Segment", key=...)
```

Every time Pandas executes:

```python
pd.Categorical(...)
```

again.

With `pd.Categorical()`:

```python
df["Segment"] = pd.Categorical(...)
```

it is created **once**, and every later sort simply uses the stored category codes.

---

## Performance

### `pd.Categorical()`

```
Create categorical once
        ↓
Sort 1 ✓
Sort 2 ✓
Sort 3 ✓
GroupBy ✓
Plot ✓
```

### `key=`

```
Sort
 ↓
Create temporary categorical
 ↓
Discard

Sort again
 ↓
Create temporary categorical
 ↓
Discard
```

So `pd.Categorical()` is more efficient if you'll reuse the ordering.

---

## Which should you use?

### Use `sort_values(key=...)` when:

* You need a **one-off custom sort**.
* You don't want to modify the DataFrame.

Example:

```python
df.sort_values(
    "Month",
    key=lambda x: pd.Categorical(
        x,
        ["Jan", "Feb", "Mar", "Apr"],
        ordered=True
    )
)
```

---

### Use `pd.Categorical()` when:

* The order is part of the meaning of the data.
* You'll sort, group, filter, or plot using that order multiple times.
* You want the DataFrame to "remember" the category order.

---

### Rule of thumb

* **One-time custom sort →** `sort_values(key=...)`
* **Persistent business/category order →** `pd.Categorical(..., ordered=True)`

This is also the convention you'll commonly see in production Pandas code.
