subset (Optional): A column label or list of labels (e.g., subset=['email']).

 Only checks the specified columns for duplicates instead of the whole row.
 
 keep (Optional): Controls which duplicate values to mark:'first' (Default):Marks all duplicates as True except for their first occurrence.
 
 keep='last': Marks all duplicates as True except for their last occurrence.
 
 keep-False: Marks all occurrences of duplicate rows as True.# Shows every row that has a duplicate somewhere in the dataset
all_duplicates = df[df.duplicated(keep=False)]


| Task                                        | Method                          |
| ------------------------------------------- | ------------------------------- |
| Remove duplicate rows                       | `drop_duplicates()`             |
| Find duplicate rows                         | `duplicated()`                  |
| Get only duplicate rows                     | `df[df.duplicated()]`           |
| Get all rows involved in duplicates         | `df[df.duplicated(keep=False)]` |
| Remove duplicates based on specific columns | `drop_duplicates(subset=[...])` |
| Detect duplicates based on specific columns | `duplicated(subset=[...])`      |


Rule of thumb:

Use duplicated() when you need a Boolean mask to inspect or filter duplicates.
Use drop_duplicates() when you want a cleaned DataFrame with duplicates removed.
