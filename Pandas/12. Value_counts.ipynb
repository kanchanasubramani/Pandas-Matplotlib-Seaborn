# **Learning guide:&#x20;**`checkout['browser'].value_counts(normalize=True)`

This Pandas expression is used to **find the percentage or proportion of records belonging to each browser**.

Let's learn it step by step.

## **1. Start with a sample dataset**

Imagine you have a DataFrame called `checkout`:

| customer_id | browser |
| ----------- | ------- |
| 1           | Chrome  |
| 2           | Safari  |
| 3           | Chrome  |
| 4           | Firefox |
| 5           | Chrome  |
| 6           | Safari  |
| 7           | Chrome  |
| 8           | Firefox |
| 9           | Chrome  |
| 10          | Safari  |

## **2. Understand the expression**

```
checkout['browser'].value_counts(normalize=True)
```

| Code              | Meaning          | What it does                                  |
| ----------------- | ---------------- | --------------------------------------------- |
| `checkout`        | DataFrame        | Refers to your dataset                        |
| `['browser']`     | Column selection | Selects only the browser column               |
| `.value_counts()` | Frequency count  | Counts how many times each browser appears    |
| `normalize=True`  | Proportion       | Converts counts into proportions of the total |

## **3. Understand the output**

```
Chrome     0.5Safari     0.3Firefox    0.2
```

| Browser   | Count  | Proportion | Percentage |
| --------- | ------ | ---------- | ---------- |
| Chrome    | 5      | 0.50       | 50%        |
| Safari    | 3      | 0.30       | 30%        |
| Firefox   | 2      | 0.20       | 20%        |
| **Total** | **10** | **1.00**   | **100%**   |

**How is the proportion calculated?**

\\[ \text{Proportion}=\frac{\text{Browser count}}{\text{Total records}} \\]

For Chrome:

\\[ \frac{5}{10}=0.50=50\\% \\]

## **4. When do we use it?**

Use `value_counts(normalize=True)` when you want to:

| Business situation  | Why use it?                                        |
| ------------------- | -------------------------------------------------- |
| Website analytics   | Find the percentage of users using each browser    |
| Customer analysis   | Understand the distribution of customer categories |
| Marketing analysis  | Find the percentage of customers in each segment   |
| Data analysis       | Understand categorical column distribution         |
| Data quality checks | Identify unexpected or dominant categories         |

For example, if Chrome accounts for 90% of website visits, you may want to prioritize testing your website on Chrome.

## **5. Compare different variations**

| Code                                                     | What you get              | Example              |
| -------------------------------------------------------- | ------------------------- | -------------------- |
| `checkout['browser'].value_counts()`                     | Actual counts             | Chrome: 5            |
| `checkout['browser'].value_counts(normalize=True)`       | Proportions               | Chrome: 0.50         |
| `checkout['browser'].value_counts(normalize=True) * 100` | Percentages               | Chrome: 50%          |
| `checkout['browser'].value_counts(ascending=True)`       | Counts in ascending order | Least frequent first |
| `checkout['browser'].value_counts(dropna=False)`         | Includes missing values   | Counts NaN too       |
