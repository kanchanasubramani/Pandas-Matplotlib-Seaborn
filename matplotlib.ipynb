### What is `rcParams`?

`rcParams` stands for **Runtime Configuration Parameters**.

It is essentially a **dictionary of default settings** that Matplotlib consults whenever it creates a figure.

For example, instead of specifying:

* figure size
* font size
* line width
* grid visibility
* DPI
* colours

every time you make a plot, you can set them once in `rcParams`.

---

## Without rcParams

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(x, y, linewidth=2)
plt.grid(True)
plt.title("Sales", fontsize=18)
```

You repeat these settings for every figure.

---

## With rcParams

```python
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["lines.linewidth"] = 2
plt.rcParams["axes.grid"] = True
plt.rcParams["axes.titlesize"] = 18

plt.plot(x, y)
plt.title("Sales")
```

Every subsequent plot automatically uses those defaults.

---

# When is rcParams used?

Matplotlib checks `rcParams` **whenever a new plotting object is created**.

For example:

```python
plt.plot(...)
```

Internally it looks up things like

```
lines.linewidth
lines.color
lines.linestyle
```

Similarly,

```python
plt.figure()
```

uses

```
figure.figsize
figure.dpi
figure.facecolor
```

and

```python
plt.xlabel()
```

uses font-related parameters.

You don't usually access `rcParams` during plotting—it influences the defaults behind the scenes.

---

# Common rcParams

```python
plt.rcParams["figure.figsize"] = (8,5)
plt.rcParams["figure.dpi"] = 120

plt.rcParams["font.size"] = 14

plt.rcParams["axes.grid"] = True
plt.rcParams["axes.titlesize"] = 18
plt.rcParams["axes.labelsize"] = 14

plt.rcParams["lines.linewidth"] = 2

plt.rcParams["legend.fontsize"] = 12
```

---

# Temporary changes

Instead of changing settings globally, you can apply them only within a block:

```python
import matplotlib as mpl

with mpl.rc_context({"font.size": 18}):
    plt.plot(x, y)
```

Outside the `with` block, the original settings are restored. This is useful in libraries or notebooks where you don't want permanent changes. ([Matplotlib][1])

---

# Difference between rcParams and style sheets

### rcParams

You change individual settings.

```python
plt.rcParams["font.size"] = 16
plt.rcParams["axes.grid"] = True
```

---

### Style sheets

You load an entire theme.

```python
plt.style.use("ggplot")
```

or

```python
plt.style.use("dark_background")
```

Internally, style sheets are largely implemented by updating collections of `rcParams`. ([Matplotlib][2])

---

# Is rcParams still relevant today?

Absolutely.

Even though higher-level libraries like **Seaborn**, **Pandas**, and **Plotnine** provide their own styling, they ultimately render through Matplotlib and often modify `rcParams` themselves.

If you want:

* consistent plots across a project,
* publication-quality figures,
* company or journal styling,
* reusable plotting defaults,

`rcParams` remains the recommended approach.

---

# Typical usage in modern code

Many projects start with a configuration cell like this:

```python
import matplotlib.pyplot as plt

plt.rcParams.update({
    "figure.figsize": (8, 5),
    "font.size": 12,
    "axes.grid": True,
    "lines.linewidth": 2,
    "figure.dpi": 120
})
```

Then all subsequent plots inherit those defaults.

---

### Should you learn `rcParams`?

Yes. If you're learning Matplotlib today, it's worth understanding because it helps you avoid repetitive styling code and makes your plots consistent. Once you're comfortable with the basics (`plt.plot`, axes, figures), learning `rcParams` is a natural next step, especially if you create many plots or want a consistent visual style across notebooks or applications.

# linewidth and marker size

-linewidth denotes the width of the line by using lw property

-markersize denotes the size of the marker by using ms property

````python
x = np.arange(1,6)
plt.plot(x,x**3,'*-g',lw=3,ms=25,mfc='yellow', alpha=0.7) 
# mfc='yellow' → fills the star marker with yellow
#alpha = transparency
plt.title('Square Function Line Plot')
plt.xlabel('N value')
plt.ylabel('Square of N')
plt.show()
````

mfc='yellow',   # Marker face color

mec='red',      # Marker edge color

mew=3           # Marker edge width

````python
plt.plot(x,x**3,'*-g',lw=3,ms=25,mfc='yellow', mec='red', mew=2) # mec='red' → marker edge color, mew=5 → marker edge width
````

| Parameter           | Meaning                                         |
| ------------------- | ----------------------------------------------- |
| `ms`                | Marker size                                     |
| `mfc`               | Marker face color (inside)                      |
| `mec`               | Marker edge color (border)                      |
| `mew`               | Marker edge width                               |
| `marker`            | Marker shape (`'o'`, `'*'`, `'s'`, `'^'`, etc.) |
| `linestyle` or `ls` | Line style (`'-'`, `'--'`, `':'`, `'-.'`)       |
| `linewidth` or `lw` | Line width                                      |
