The error occurs because you've passed the arguments to `np.linalg.solve()` in the wrong order.

The syntax is:

```python
np.linalg.solve(A, B)
```

where:

* `A` is the **coefficient matrix** (must be a square 2D matrix).
* `B` is the **constant vector** (or matrix).

In your code:

```python
import numpy as np

a = [2200, 10100]          # Constant vector
b = [[1, 1], [3, 8]]       # Coefficient matrix

np.linalg.solve(a, b)      # Incorrect
```

Here, `a` is a 1D array, but NumPy expects the first argument to be a 2D square matrix. That's why you get:

```
LinAlgError: 1-dimensional array given. Array must be at least two-dimensional
```

### Correct code

```python
import numpy as np

A = [[1, 1],
     [3, 8]]

B = [2200, 10100]

x = np.linalg.solve(A, B)
print(x)
```

**Output:**

```python
[1000. 1200.]
```

This means the solution to the system

[
x + y = 2200
]

[
3x + 8y = 10100
]

is:

* (x = 1000)
* (y = 1200)

So, simply swap the arguments:

```python
np.linalg.solve(b, a)
```

or, more clearly, name them as `A` and `B` to avoid confusion.


````python
# dot product


def dot_product(a,b):
    result=0
    for i,j in zip(a,b):
        result=result+i*j
    return result
````

````python
a=np.array([10,20,30])
b=np.array([1,2,3])
x=dot_product(a,b)
print(x)
````

````python
a=np.array([10,20,30])
b=np.array([1,2,3])
c=np.array([4,5,6])
d=np.array([7,8,9])
````

````python
def dot_product(*args):
    result=0
    for i in zip(*args):
        result+=np.prod(i)
    return result
````

````python
dot_product(a,b,c,d)
````
